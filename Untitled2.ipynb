```python
import spacy
```

```python
nlp=spacy.blank("en")
doc=nlp("Captain America ate 100$ samosa. Then he said I can do this al day.")
for  token in doc:
    print(token)
```

```python
nlp.pipe_names
```

```python
npl=spacy.load("en_core_web_sm")
```

```python
npl.pipeline
```

```python
doc=npl("Captain America ate 100$ samosa. Then he said I can do this al day.")

for token in doc:
    print(token, "|", token.pos_, "|", token.lemma_)
```

```python
npl.pipe_names
```

```python
doc=npl("Tesla Inc is going to acquire twitter for $45 billion")

for ent in doc.ents:
    print(ent.text,"|",ent.label_)
```
